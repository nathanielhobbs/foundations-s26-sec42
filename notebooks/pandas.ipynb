# Loading data

Pandas works with "data frames", which can be thought of as a table with rows and columns.

You can load data from many different kinds of file types: csv, xls, html, xml, etc.

```python
df = pd.read_csv('practice_data.csv')
```

In Jupyter notebook in order to see the data frame, you can call the variable by itslef

```python
df
```

Some useful commands for seeing a subset of rows of the data frame are 

- head (for seeing the beginning of the data frame)
- tail (for seeing the end of the data frame)

```python
df.head()
```

```python
df.tail(7)
```

### Working with Columns

You can also view particular columns of the data frame  by indexing the name of the column

```python
df['Name'] # get the entire column
```

```python
df['Name'][4:9] # get rows 4 through 8 of the column
```

You can get multiple columns from a dataframe by indexing a *list* of column names

```python
df[['Name', 'Type 1']][4:9]
```

### Working with rows

besides head() and tail() you can get any subset of rows using iloc(), which stands for integer location

```python
# get rows 20 to 25
df.iloc[20:26]
```

### Getting a particluar cell (by row/column)

df.iloc[rows, cols]

```python
df.iloc[20:26, 5:8]
```

You can also iterate through each row in a loop

```python
for idx,row in df.iterrows():
    print(idx,row)
```

You can get a subset of rows according to some Boolean condition using the .loc() function

```python
df.loc[df['Type 1'] == 'Fire']
```

#### Filtering data 

Again, you can filter which rows you are working with using Boolean criteria with the .loc() method.

```python
df.loc[df['Type 1'] == 'Fire']
```

You can filter by multiple critera by chaining together Boolean expressions.

Note, in pandas these quantifiers:

& is used for AND

| is used for OR

~ is used for NOT

```python
df.loc[(df['Type 1'] == 'Fire') & (df['Type 2'] == 'Dragon')]
```

```python
df.loc[(df['Type 1'] == 'Fire') | (df['Type 2'] == 'Dragon')]
```

```python
df.loc[~(df['Type 1'] == 'Fire')]
```

Above, the indices are "all over the place".  However, you can reset them so that they range from 0 to the length of the dataframe

```python
my_df = df.loc[~(df['Type 1'] == 'Fire')]

# my_df.drop(columns='index')

my_df.reset_index()
```
